```python
# imports 
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import openpyxl
```

```python
DATA_FILE = Path("data/mwd_bet_aligned_features.xlsx")
SITE_SHEETS = [f'B{i}_DATA' for i in range(8, 16)]

all_sites = []
for sheet in SITE_SHEETS:
    df = pd.read_excel(DATA_FILE, sheet_name=sheet, engine='openpyxl')
    df['SiteID'] = sheet.replace('_DATA', '')
    all_sites.append(df)

df_master = pd.concat(all_sites, ignore_index=True)
print(f"Loaded: {len(df_master)} rows × {len(df_master.columns)} columns")
interpolated_cols_mwd = ['AS', 'HP', 'IF', 'IP', 'RSP', 'TP', 'TQ', 'TPAF', 'TQAT']
df_master = df_master.drop(columns=interpolated_cols_mwd)
df_master.head()
```
